Data Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

x = df[["bedrooms", "bathrooms", "sqft_living", "sqft_lot", "floors", "condition", "sqft_above", "sqft_basement"]]
y = df["price"]

features = ["bedrooms", "bathrooms", "sqft_living", "sqft_lot", "floors", "condition", "sqft_above", "sqft_basement"]
#Plot Actual Data
# plt.scatter(x, y)

# Fit Linear Regression Line (y = mx + c)
# m, c = np.polyfit(x, y, 1)

# Predict values
# y_pred = m * x + c

# Plot with Predicted
# plt.plot(x, y_pred)

# Titles
for feature in features:

  X = df[feature]
  m, c = np.polyfit(X, y, 1)
  y_pred = m * X + c

  plt.scatter(X, y)
  plt.plot(X, y_pred)
  plt.xlabel("x Values")
  plt.ylabel("y Values (Price)")
  plt.title(f"Linear Regression for {feature}")

  plt.show()
```

#Model Training

```python
from sklearn.linear_model import LinearRegression
```

```python
x = df.iloc[:, 1:]
y = df.price
```

```python
df.head()
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
x_train.head()
```

```python
x_train.bedrooms.unique()
```

```python
model = LinearRegression()

model.fit(x_train, y_train)
```

```python
model.score(x_train, y_train)
```

```python
model.score(x_test, y_test)
```

```python
x_train_copy = x_train.copy()
y_train_copy = y_train.copy()

x_test_copy  = x_test.copy()
y_test_copy  = y_test.copy()
```

```python
model.coef_
```

```python
model.intercept_
```

Error Calculation

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, root_mean_squared_error, mean_absolute_percentage_error, r2_score
from sklearn.model_selection import cross_val_score
```

```python
y_pred_train = model.predict(x_train)
y_pred_test = model.predict(x_test)
```

Mean Squared Error

```python
# Mean Squared Error
# On Train
mse = mean_squared_error(y_train, y_pred_train)
print("mse", mse)

print("Score", r2_score(y_train, y_pred_train))
```

```python
mse_test = mean_squared_error(y_test, y_pred_test)
print("mse_test", mse_test)

print("Score", r2_score(y_test, y_pred_test))
```

Mean Absolute Error

```python
mae = mean_absolute_error(y_train, y_pred_train)
print("mae", mae)

mae_test = mean_absolute_error(y_test, y_pred_test)
print("mae_test", mae_test)
```

Root Mean Squared Error

```python
rmse = np.sqrt(root_mean_squared_error(y_train, y_pred_train))
print(rmse)

rmse_test = np.sqrt(root_mean_squared_error(y_test, y_pred_test))
print(rmse_test)
```

Mean Absolute Percentage Error

```python
mape = mean_absolute_percentage_error(y_train, y_pred_train)
print(mape)

mape_test = mean_absolute_percentage_error(y_test, y_pred_test)
print(mape_test)
```
